```python
import json
import csv
import requests
from bs4 import BeautifulSoup
from fake_useragent import UserAgent
```

```python
target_url = 'https://www.scrapethissite.com/pages/forms/'
with open('hockey_team.csv','w',encoding='utf-8',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Team Name','Year','Wins','Losses','win %','Goals for','Goals Against'])
    while target_url:
        ua = UserAgent()
    
        headers = {
            'User-agent': ua.random,
            'Accept-Language':'en-US,en;q=0.9',
            'Accept-Encodin':'gzip, deflate, br, zstd'
        }
        response = requests.get(target_url,headers=headers)
        raw_html = response.text
        soup = BeautifulSoup(raw_html,'html.parser')
        table = soup.find('table',class_='table')
        teams = table.find_all('tr',class_='team')

        for team in teams:
            team_name = team.find('td',class_='name')
            year = team.find('td',class_='year')
            wins = team.find('td',class_='wins')
            losses = team.find('td',class_='losses')
            percentage_winning = team.find('td',class_='pct')
            goals_for = team.find('td',class_='gf')
            goals_against = team.find('td',class_='ga')
            writer.writerow([team_name.text.strip(),year.text.strip(),wins.text.strip(),losses.text.strip(),percentage_winning.text.strip(),goals_for.text.strip(),goals_against.text.strip()])
        next_page = soup.find('a',attrs={'aria-label':'Next'})
        if next_page:
            link_url = next_page['href']
            target_url = 'https://www.scrapethissite.com' + link_url
        else:
            target_url = None
            break
    

print('I did it')
```
